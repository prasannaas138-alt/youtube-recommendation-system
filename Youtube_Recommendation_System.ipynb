```python
text_columns = ['title','description','tags','category']

for column in text_columns:
  df[column] = df[column].fillna("").astype(str)
```

```python
df['combined_text'] = (
    df['title'] + " " +
    df['description'] + " " +
    df['tags'] + " " +
    df['category']
)
```

```python
df['combined_text'] = df['combined_text'].str.lower()
```

```python
df['combined_text'] = df['combined_text'].str.replace(r"\s+" , " " , regex = True).str.strip()
```

```python
df[['title' , 'combined_text']].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(df['combined_text'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_matrix = cosine_similarity(tfidf_matrix)
```

```python
similarity_matrix.shape
```

```python
similarity_scores = similarity_matrix[0]
```

```python
similar_indeces = similarity_scores.argsort()[::-1][:6]
```

```python
df.iloc[similar_indeces][['video_id' , 'title' , 'category']]
```

```python
def recommend_video (title , n=5):
  index = df[df['title'] == title].index[0]

  similarity_scores = similarity_matrix[index]

  similarity_indices = similarity_scores.argsort()[::-1][1:n+1]

  recommendation = df.iloc[similarity_indices].copy()

  recommendation['similarity_score'] = similarity_scores[similarity_indices]

  return recommendation[
      ['video_id' , 'title' , 'category' , 'channel' , 'similarity_score']
  ]
```

```python
def search_video(query , n=10 , threshold=0.10):

  query_vector = tfidf.transform([query])

  similarity_scores = cosine_similarity(query_vector , tfidf_matrix).flatten()

  similarity_indices = similarity_scores.argsort()[::-1]

  top_indices = similarity_indices[:n]

  if similarity_scores[top_indices[0]] < threshold:
    print('Video does not exist')
    return

  recommendation = df.iloc[top_indices].copy()

  recommendation['similarity_score'] = similarity_scores[top_indices]

  return recommendation[
      ['video_id' , 'title' , 'category'  , 'channel' , 'similarity_score']
  ]
```
